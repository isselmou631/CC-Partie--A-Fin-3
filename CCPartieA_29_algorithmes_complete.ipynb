## 1. Régression linéaire

Construire le modèle, afficher une métrique ou une sortie et interpréter ses limites. Les exemples n’utilisent pas de test final pour le réglage d’hyperparamètres.



### Présentation

**Famille et objectif.** Régression linéaire appartient à la famille **Régression**. Il vise à prédire une variable quantitative à partir d'une combinaison linéaire des variables explicatives.

**Intuition et étapes.** Le modèle choisit la droite ou l'hyperplan qui minimise la somme des carrés des résidus. Entrée X → estimation des coefficients → prédiction ŷ. ŷᵢ = β₀ + Σⱼ βⱼxᵢⱼ ; β sont les paramètres appris, x les variables et β₀ l'ordonnée à l'origine. Il minimise Σᵢ(yᵢ−ŷᵢ)².

**Avantages.** Simple et très interprétable ; rapide et efficace quand la relation est approximativement linéaire.  
**Limites.** Sensible aux valeurs extrêmes et à la multicolinéarité ; ne représente pas bien les relations fortement non linéaires.

## 2. Ridge

Construire le modèle, afficher une métrique ou une sortie et interpréter ses limites. Les exemples n’utilisent pas de test final pour le réglage d’hyperparamètres.



### Présentation

**Famille et objectif.** Ridge appartient à la famille **Régression régularisée**. Il vise à stabiliser une régression linéaire en pénalisant les coefficients de grande amplitude.

**Intuition et étapes.** Après standardisation, Ridge minimise l'erreur quadratique augmentée d'une pénalité L2. Entrée X → compromis ajustement/simplicité → prédiction. minβ [Σᵢ(yᵢ−xᵢᵀβ)² + αΣⱼβⱼ²] ; α est un hyperparamètre, β les paramètres appris.

**Avantages.** Réduit la variance et l'effet de la multicolinéarité ; conserve toutes les variables et reste interprétable.  
**Limites.** Ne sélectionne pas réellement les variables ; le choix de α influence fortement le biais et doit être validé sans toucher au test.

### Schéma explicatif

Pour les méthodes séquentielles, la sortie/récompense revient dans la boucle d'apprentissage. **Données et prérequis.** Cible continue, variables numériques encodées et standardisation recommandée ; validation croisée pour α.

**Cas peu approprié.** Peu appropriée si des variables inutiles doivent être supprimées automatiquement ou si la relation est très non linéaire.

**Métriques.** MAE, RMSE, R² et stabilité des coefficients entre échantillons.

### Présentation

**Famille et objectif.** Lasso appartient à la famille **Régression régularisée Lasso**. Il vise à prédire une quantité tout en produisant une sélection parcimonieuse de variables.

**Intuition et étapes.** Lasso minimise l'erreur et une pénalité L1 qui peut annuler certains coefficients. X → estimation sparse → ŷ. minβ [Σᵢ(yᵢ−xᵢᵀβ)² + αΣⱼ|βⱼ|] ; α est l'hyperparamètre de régularisation.

**Avantages.** Sélection automatique de variables ; modèle relativement interprétable et utile en haute dimension.  
**Limites.** Sélection instable avec des variables très corrélées ; peut sous-ajuster si α est trop grand et dépend de l'échelle des variables.

**Cas peu approprié.** Peu appropriée lorsque toutes les variables ont un effet faible mais utile et sont fortement corrélées.

Construire le modèle, afficher une métrique ou une sortie et interpréter ses limites. Les exemples n’utilisent pas de test final pour le réglage d’hyperparamètres.



### Présentation

**Famille et objectif.** SARSA appartient à la famille **SARSA**. Il vise à apprendre une valeur d'action selon la politique effectivement suivie, donc une méthode on-policy.

**Intuition et étapes.** L'agent choisit a, observe r et s', choisit a' selon la même politique, puis met à jour vers Q(s',a'). Q←Q+α[r+γQ(s',a')−Q(s,a)]; α, γ et ε sont des hyperparamètres.

**Avantages.** Tient compte du risque de la politique exploratoire ; simple et souvent plus prudent que Q-learning.  
**Limites.** Dépend de la politique d'exploration ; apprentissage plus lent et limité aux espaces discrets tabulaires.

````python
import gymnasium as gym
env=gym.make('FrozenLake-v1',is_slippery=False);Q=np.zeros((16,4));generateur=np.random.default_rng(GRAINE)
def choisir(etat,epsilon):
    return int(generateur.integers(4)) if generateur.random()<epsilon else int(np.argmax(Q[etat]))
for episode in range(4000):
    etat,_=env.reset(seed=GRAINE+episode);epsilon=max(.05,1-episode/2500);action=choisir(etat,epsilon)
    for pas in range(100):
        suivant,recompense,termine,tronque,_=env.step(action)
        action_suiv=choisir(suivant,epsilon)
        cible=recompense+(0 if termine else .95*Q[suivant,action_suiv])
        Q[etat,action]+=.3*(cible-Q[etat,action]);etat,action=suivant,action_suiv
        if termine or tronque:break
print('Politique SARSA:',Q.argmax(axis=1));env.close()
````
